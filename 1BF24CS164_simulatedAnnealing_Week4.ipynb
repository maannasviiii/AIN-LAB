```python
import random
import math

def function(x):
    return x * x

current = random.uniform(-10, 10)
best = current

temperature = 1000
cooling_rate = 0.95

while temperature > 0.01:

    next = current + random.uniform(-1, 1)

    delta = function(next) - function(current)

    if delta < 0:
      current = next
    else:
      probability = math.exp(-delta / temperature)

      if random.random() < probability:
        current = next

    if function(current) < function(best):
      best = current

    temperature = temperature * cooling_rate

print("Best solution:", best)
print("Minimum value:", function(best))
```
