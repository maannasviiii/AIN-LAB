```python
def unify(x, y, subst):
    if x == y:
        return subst

    if x.startswith("?"):
        if x in subst:
            return unify(subst[x], y, subst)
        if y == x:
            return subst
        subst[x] = y
        return subst

    if y.startswith("?"):
        return unify(y, x, subst)

    return None


def main():
    p1 = input("Enter first predicate: ")
    p2 = input("Enter second predicate: ")

    name1 = p1[:p1.index("(")]
    name2 = p2[:p2.index("(")]

    args1 = p1[p1.index("(") + 1:p1.index(")")].split(",")
    args2 = p2[p2.index("(") + 1:p2.index(")")].split(",")

    args1 = [x.strip() for x in args1]
    args2 = [x.strip() for x in args2]

    if name1 != name2 or len(args1) != len(args2):
        print("Unification Failed")
        return

    subst = {}

    for x, y in zip(args1, args2):
        subst = unify(x, y, subst)
        if subst is None:
            print("Unification Failed")
            return

    print("Unification Successful")
    print("Substitution:", subst)


main()
```
