```python
class ReflexVacuumAgent:
    def __init__(self):
        # The environment state: 'Clean' or 'Dirty'
        self.environment = {'A': 'Dirty', 'B': 'Dirty'}
        self.robot_location = 'A'  # Starts in Room A

    def check_percept(self):
        """Returns the current location and its status."""
        return self.robot_location, self.environment[self.robot_location]

    def act(self):
        """Executes an action based on the current room status."""
        location, status = self.check_percept()
        print(f"Robot is in Room {location}. Status: {status}")

        if status == 'Dirty':
            print(f"↳ Action: Sucking dirt in Room {location}...")
            self.environment[location] = 'Clean'
        else:
            # If current room is clean, move to the opposite room
            new_location = 'B' if location == 'A' else 'A'
            print(f"↳ Action: Room {location} is clean. Moving to Room {new_location}...")
            self.robot_location = new_location

    def run_simulation(self, steps=4):
        print("--- Starting Reflex Vacuum Simulation ---")
        for step in range(1, steps + 1):
            print(f"\n[Step {step}]")
            self.act()
        print("\nSimulation complete. Final Environment:", self.environment)

# Run the reflex agent
agent = ReflexVacuumAgent()
agent.run_simulation()
```
