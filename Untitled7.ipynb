```python
def get_neighbors(state):
    neighbors = []
    zero_index = state.index(0)
    row, col = zero_index // 3, zero_index % 3

    moves = [(0, 1), (0, -1), (1, 0), (-1, 0)]

    for dr, dc in moves:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero_index = new_row * 3 + new_col
            new_state_list = list(state)
            new_state_list[zero_index], new_state_list[new_zero_index] = \
                new_state_list[new_zero_index], new_state_list[zero_index]
            neighbors.append(tuple(new_state_list))
    return neighbors

def depth_limited_search(state, goal, depth, path):

    if state == goal:
        return path

    if depth == 0:
        return None

    for neighbor in get_neighbors(state):

        if neighbor not in path:

            result = depth_limited_search(
                neighbor,
                goal,
                depth - 1,
                path + [neighbor]
            )

            if result is not None:
                return result

    return None


def ids(initial, goal, max_depth=20):

    depth = 0

    while True:
        if max_depth is not None and depth > max_depth:
            print(f"Search reached max_depth: {max_depth}. No solution found within this limit.")
            return None

        result = depth_limited_search(
            initial,
            goal,
            depth,
            [initial]
        )

        if result is not None:
            return result

        depth += 1


def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


initial = (2, 8, 3,
           1, 6, 4,
           7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)

print("Time Complexity: O(b^d), where b is branching factor, d is depth of solution")
print("Space Complexity: O(bd), where b is branching factor, d is depth of solution")

while True:
    try:
        user_max_depth = int(input("Enter the maximum depth for the search : "))
        if user_max_depth >= 0:
            break
        else:
            print("Please enter a non-negative integer.")
    except ValueError:
        print("Invalid input. Please enter an integer.")

solution = ids(initial, goal, max_depth=user_max_depth)

if solution:
    print("\nIDS Solution:")

    for state in solution:
        print_state(state)
else:
    print("\nNo solution found within the specified maximum depth.")
```
