```python
from itertools import product

# Knowledge Base in CNF form
# P -> Q  =  (~P OR Q)
# Q -> R  =  (~Q OR R)
# P is True

kb = [
    ["~P", "Q"],
    ["~Q", "R"],
    ["P"]
]

query = input("Enter query (P, Q or R): ")
symbols = ["P", "Q", "R"]

def evaluate(clause, values):
    for literal in clause:
        if literal.startswith("~"):
            if not values[literal[1:]]:
                return True
        else:
            if values[literal]:
                return True
    return False

def check_kb(values):
    for clause in kb:
        if not evaluate(clause, values):
            return False
    return True

def entails():
    for combination in product([True, False], repeat=3):
        values = dict(zip(symbols, combination))
        if check_kb(values):
            if not values[query]:
                return False
    return True

if query not in symbols:
    print("Invalid Query")
elif entails():
    print("Knowledge Base entails the query")
else:
    print("Knowledge Base does not entail the query")
```
